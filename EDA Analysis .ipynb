# Bivariate Analysis

```python
sns.scatterplot(data=df,x='Age',y='Fare',alpha=0.6) # alpha does the transparency 
plt.show()
```

```python
print(df[['Age','Fare']].corr())
print('correalation of age in fare is ',df['Age'].corr(df['Fare']))
```

```python
fig , axes = plt.subplots(1,2,figsize=(12,4))
sns.boxplot(data=df,x='Pclass',y='Fare',ax=axes[0])
axes[0].set_title('Fare Distribution by Passenger Class')
sns.violinplot(data=df,x='Pclass',y='Fare',ax=axes[1])
axes[1].set_title('Fare distirbution by passenger class ')
plt.tight_layout()
plt.show()
```

```python
df.groupby('Pclass')['Fare'].agg(['mean','median','min','max']) 
```

```python
sns.boxplot(data=df,x='Pclass',y='Age')
plt.show()
```

```python
df.groupby('Pclass')['Age'].agg(['mean','median','min','max'])
```

```python
# sex vs fare 
sns.boxplot(data=df,x='Sex',y='Fare')
plt.show()
```

```python
df.groupby('Sex')['Fare'].agg(['mean','median','min','max'])  
```

```python
fig , ax = plt.subplots(1,2,figsize=(12,4))
sns.violinplot(data=df, x='Survived', y='Age',ax=ax[0])

sns.boxplot(data=df,x='Survived',y='Age',ax=ax[1])
plt.show()
```

```python
df.groupby('Survived')['Age'].agg(['mean','median','min','max'])
```

```python
print(pd.crosstab(df['Pclass'], df['Survived']))
pd.crosstab(
    df['Pclass'],
    df['Survived'],
    normalize='index'
)*100
```

```python
print(pd.crosstab(df['Survived'], df['Sex']))
pd.crosstab(
    df['Survived'],
    df['Sex'],
    normalize='index'
)*100
```

```python
sns.countplot(data=df,x='Sex',hue='Survived')
```

```python
print('---------------------------------')
print(pd.crosstab(df['Pclass'],df['Survived']))
print('---------------------------------')

print(pd.crosstab(
    df['Survived'],
    df['Sex'],
    normalize='index'
)*100)
print('----------------------------------')
sns.countplot(data=df,x='Pclass',hue='Survived')
```

```python
print('---------------------------------')
print(pd.crosstab(df['Embarked'],df['Survived']))
print('---------------------------------')

print(pd.crosstab(
    df['Embarked'],
    df['Survived'],
    normalize='index'
)*100)
print('----------------------------------')
sns.countplot(data=df,x='Embarked',hue='Survived')
```
